```python
def angle_to_pulse_us(angle, a_min=-90, a_max=90, p_min=1000, p_max=2000):
    angle = max(a_min, min(a_max, angle)) # clamp first
    return p_min + (angle - a_min) / (a_max - a_min) * (p_max - p_min)
```

Update rate: a standard analog servo only accepts a new command every 20 ms (50 Hz). Even if your PID
runs at 500 Hz, the servo only acts on every 10th output. Digital servos accept faster rates (often 200–
333 Hz).
Servo speed: specs like “0.10 s / 60°” mean it can’t move faster than that. This is a rate limit, a
nonlinearity your PID must live with.
Deadband: servos ignore tiny changes (a few µs of pulse width), so very small corrections don’t happen.
