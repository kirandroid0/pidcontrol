```python
import numpy as np
import matplotlib.pyplot as plt
import control as ct

s = ct.tf("s") # lets you write transfer functions naturally
# Lesson 2's plant and PI controller
K, tau = 2.0, 5.0
P = K / (tau * s + 1)
C = 2.0 + 0.4 / s # Kp = 2, Ki = 0.4 (Ti = tau)
L = C * P # loop transfer function
T = ct.feedback(L, 1) # closed loop Y/R = L/(1+L)
print("Closed loop before cancellation:", T)
print("Poles:", ct.poles(T), " Zeros:", ct.zeros(T))
print("Minimal realization:", ct.minreal(T, verbose=False)) # cancels the common pole/zero
t = np.linspace(0, 8, 801)
t_out, y_out = ct.step_response(T, t)
print(f"y at t = 1.25 s (one closed-loop time constant): {np.interp(1.25, t_out, y_out):.3f} (expect 0.632)")
# Disturbance transfer function Y/D = P/(1+CP)
Td = ct.feedback(P, C) # forward P, feedback C
print("DC gain of Y/D:", ct.dcgain(Td), "(zero: the integrator rejects constant disturbances)")
```
